# 🧠 Python + SQL + Tableau Integrated Data Analysis Guide


This notebook provides a professional workflow for **data analysts** who wish to combine:
- **Python** (for data cleaning & transformation)
- **SQL** (for querying structured data)
- **Tableau** (for visualization and reporting)

You’ll work through a **fintech case study** dataset with real-world issues — missing values, messy symbols, and inconsistent formats.


## 1️⃣ Setup Libraries

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
from scipy import stats

sns.set(style='whitegrid')
print("Libraries loaded successfully ✅")
```

## 2️⃣ Load & Inspect Data

```

# Load your messy fintech dataset (update file path)
df = pd.read_csv(r'C:\Users\YourName\Downloads\fintech_data.csv')

# Inspect
print(df.shape)
df.info()
df.head()
```

## 3️⃣ Clean and Transform Data

```

# Remove unwanted characters and convert numeric columns
df['loan_amount'] = df['loan_amount'].astype(str).str.replace('[₦,%]', '', regex=True).astype(float)
df['interest_rate'] = df['interest_rate'].astype(str).str.replace('%', '', regex=True).astype(float)

# Handle missing dates (forward fill)
df['start_date'] = pd.to_datetime(df['start_date'], format='%Y%m%d', errors='coerce')
df['start_date'] = df['start_date'].fillna(method='bfill')

# Handle missing numeric values with median
df['loan_amount'] = df['loan_amount'].fillna(df['loan_amount'].median())

# Drop duplicates
df = df.drop_duplicates()
```

## 4️⃣ Exploratory Data Analysis

```

sns.pairplot(df[['loan_amount', 'annual_income', 'interest_rate']], diag_kind='kde')
plt.show()

# Correlation heatmap
sns.heatmap(df[['loan_amount', 'annual_income', 'interest_rate']].corr(), annot=True, cmap='coolwarm')
plt.show()
```

## 5️⃣ Save Clean Data for SQL & Tableau

```

df.to_csv('cleaned_fintech_data.csv', index=False)
print("✅ Clean data saved as cleaned_fintech_data.csv")
```

## 6️⃣ SQL Integration Example

```

conn = sqlite3.connect('fintech.db')
df.to_sql('loans', conn, if_exists='replace', index=False)

# Example SQL query
query = "SELECT state, AVG(loan_amount) as avg_loan FROM loans GROUP BY state"
sql_df = pd.read_sql(query, conn)
conn.close()

sql_df.head()
```
